## Import necessary libraries

```python
from sklearn import datasets
import pandas as pd
```

```python
breast_cancer_dataset = datasets.load_breast_cancer()
```

```python
breast_cancer_dataset
```

```python
## To read the information of the dataset

print(breast_cancer_dataset.DESCR)
```

```python
## Make a dataframe from the given dataset

breast_cancer_df = pd.DataFrame(data=breast_cancer_dataset["data"], columns = breast_cancer_dataset["feature_names"])
```

```python
## Add the target column to the dataframe

breast_cancer_df["Target"] = breast_cancer_dataset["target"]
```

```python
breast_cancer_df
```

## 3. Data Understanding

```python
breast_cancer_df.shape
```

```python
breast_cancer_df.isnull().sum()
```

```python
breast_cancer_df.dtypes
```

## 4.Model Building

```python
X =  breast_cancer_df.drop(labels = "Target", axis = 1)
```

```python
y = breast_cancer_df["Target"]
```

```python
y.value_counts()
```

```python
from matplotlib import pyplot as plt

plt.pie(y.value_counts(), labels = ["Malignent", "Benign"], autopct='%1.0f%%')
plt.show()
```

```python
%%time

from sklearn.model_selection import train_test_split

X_train,X_test,y_train,y_test = train_test_split(X,y, test_size=0.2,random_state=123, shuffle=True, stratify=y)
X_train,X_test,y_train,y_test
```

```python
print("Train Samples:", X_train.shape,X_test.shape)
print("Test Samples:", y_train.shape,y_test.shape)
```

## 5. Model Training

```python
from sklearn.tree import DecisionTreeClassifier

dt_model = DecisionTreeClassifier(criterion='gini', max_depth= 3, class_weight={0:2,1:1})

dt_model.fit(X_train,y_train)
```

```python
y.value_counts()

from matplotlib import pyplot as plt

plt.pie(y_train.value_counts(), labels = ["Malignent", "Benign"], autopct='%1.0f%%')
plt.show()
```

## 6. Model Testing & Evaluation

### For Train

```python
y_pred_train = dt_model.predict(X_train)
y_pred_train
```

```python

from matplotlib import pyplot as plt

plt.pie(y_test.value_counts(), labels = ["Malignent", "Benign"], autopct='%1.0f%%')
plt.show()
```

```python
from sklearn.metrics import confusion_matrix, classification_report


print(confusion_matrix(y_train,y_pred_train))
```

```python
print(classification_report(y_train,y_pred_train))
```

```python
### For Testing
```

```python
y_pred_test = dt_model.predict(X_test)
y_pred_test
```

```python
print(confusion_matrix(y_test,y_pred_test))
```

```python
print(classification_report(y_test,y_pred_test))
```

## Model Optimization Techniques

```python
from sklearn.preprocessing import StandardScaler
```

```python
std = StandardScaler()
```

```python
Scaled_X = std.fit_transform(X,y)
Scaled_X
```

```python
scaled_X = pd.DataFrame(Scaled_X, columns = breast_cancer_dataset["feature_names"])

scaled_X
```

```python
scaled_X.mean(), scaled_X.std()
```

```python
%%time
from sklearn.model_selection import train_test_split

X_train,X_test,y_train,y_test = train_test_split(scaled_X,y, test_size=0.2,random_state=123, shuffle=True, stratify=y)
X_train,X_test,y_train,y_test

from sklearn.tree import DecisionTreeClassifier

dt_model = DecisionTreeClassifier(criterion='gini', max_depth= 3, class_weight={0:2,1:1})

dt_model.fit(X_train,y_train)

y_pred_train = dt_model.predict(X_train)
#y_pred_train

from sklearn.metrics import confusion_matrix, classification_report


print(confusion_matrix(y_train,y_pred_train))

print(classification_report(y_train,y_pred_train))

y_pred_test = dt_model.predict(X_test)
#y_pred_test

print(confusion_matrix(y_test,y_pred_test))

print(classification_report(y_test,y_pred_test))
```
